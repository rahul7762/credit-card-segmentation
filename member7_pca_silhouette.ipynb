**Import** **Required** **Libraries**                   
 In this step, we import the libraries required for PCA and cluster evaluation. PCA is used for dimensionality reduction and visualization, while Silhouette Score is used to evaluate the quality of the clusters.


```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
```

**Principal Component Analysis (PCA)**

PCA is applied to the scaled dataset to reduce the high-dimensional data into two principal components. These two components capture the maximum possible variance in the data and allow us to visualize the customer segments in a 2D space.

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("Original features:", X_scaled.shape[1])
print("PCA components:", X_pca.shape[1])
print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total explained variance:", pca.explained_variance_ratio_.sum())
```

**PCA Visualization**

The two principal components are plotted to visualize the customer clusters in a two-dimensional space. Each point represents a customer, and the color indicates the cluster assigned by the clustering algorithm.

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=labels,
    cmap="viridis",
    s=20
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Customer Segments using PCA")
plt.colorbar(label="Cluster")

plt.show()
```

**Silhouette Score for Agglomerative Clustering**

Silhouette Score is used to evaluate the quality of the Agglomerative Clustering results. It measures how similar a customer is to its own cluster compared with other clusters.


```python
agglo_silhouette = silhouette_score(X_scaled, labels)

print("Agglomerative Clustering Silhouette Score:",
      round(agglo_silhouette, 4))
```

**K-Means vs Agglomerative Comparison**

In this step, we compare the clustering quality of K-Means and Agglomerative Clustering using their Silhouette Scores. The algorithm with the higher score provides better-separated and more compact clusters according to this evaluation metric.

```python
kmeans_silhouette = silhouette_score(X_scaled, cluster_labels)

print("K-Means Silhouette Score:",
      round(kmeans_silhouette, 4))

print("Agglomerative Silhouette Score:",
      round(agglo_silhouette, 4))
```

**Model Comparison Visualization**

A bar chart is created to visually compare the Silhouette Scores of K-Means and Agglomerative Clustering. This makes it easier to identify which clustering algorithm achieved better cluster quality.

```python
models = ["K-Means", "Agglomerative"]
scores = [kmeans_silhouette, agglo_silhouette]

plt.figure(figsize=(7, 5))
plt.bar(models, scores)

plt.xlabel("Clustering Algorithm")
plt.ylabel("Silhouette Score")
plt.title("K-Means vs Agglomerative Clustering")

plt.show()
```
